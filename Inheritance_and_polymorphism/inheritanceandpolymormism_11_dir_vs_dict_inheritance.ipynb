```python
# Customizing with __dir__
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __dir__(self):
        # Customize what dir() returns
        return ['name', 'greet']

    def greet(self):
        return f"Hello, my name is {self.name}"

p = Person("Alice", 25)

print(dir(p))
```

```python
# Showing the difference clearly
class Demo:
    def __init__(self):
        self.x = 10
        self.y = 20

    def __dir__(self):
        return ['x']  # Only expose 'x'

d = Demo()

print("dir(d):", dir(d))
print("Actual attributes:", d.__dict__)
```

```python
class MyClass:
    class_var1 = "I'm a class variable"
    class_var2 = 42
    __internal_value = 0
    
    def __init__(self):
        self.instance_var = "I'm an instance variable"

# Access class variables through the class's __dict__
print(MyClass.__dict__)
```

```python
# Get only regular class variables (excluding methods, special attributes, etc.)
class_vars = {key: value for key, value in MyClass.__dict__.items() 
              if not key.startswith('__') and not callable(value)}
print(class_vars)
```
