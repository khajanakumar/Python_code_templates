# Multiple Inheritance in Python: Building a Multifunction Device
- When you run this code, you'll notice different behavior for the `print()` method between the two classes. This demonstrates a key aspect of multiple inheritance:

    - 1. In `MFD_SPF` (Scanner, Printer, Fax), the `print()` method will come from the `Printer` class because it appears before `Fax` in the inheritance order.

    - 2. In `MFD_SFP` (Scanner, Fax, Printer), the `print()` method will come from the `Fax` class because it appears before `Printer` in the inheritance order.

    - This illustrates Python's Method Resolution Order (MRO) algorithm, which determines which method to call when multiple parent classes define the same method. Python uses the C3 linearization algorithm to determine the order in which it searches for methods in the inheritance hierarchy.

    - The output of the MRO at the end shows the exact search path Python uses when looking for methods, which helps explain why the `print()` method behaves differently in each class.

```python
class Scanner:
    def scan(self):
        print('scan() method from Scanner class')

class Printer:
    def print(self):
        print('print() method from Printer class')

class Fax:
    def send(self):
        print('send() method from Fax class')
    
    def print(self):
        print('print() method from Fax class')

# First multifunction device: Scanner -> Printer -> Fax order
class MFD_SPF(Scanner, Printer, Fax):
    pass

# Second multifunction device: Scanner -> Fax -> Printer order
class MFD_SFP(Scanner, Fax, Printer):
    pass

# Create instances of both classes
device_spf = MFD_SPF()
device_sfp = MFD_SFP()

# Test the first device (SPF)
print("Testing MFD_SPF device:")
device_spf.scan()
device_spf.print()
device_spf.send()

print("\n" + "-" * 40 + "\n")

# Test the second device (SFP)
print("Testing MFD_SFP device:")
device_sfp.scan()
device_sfp.print()
device_sfp.send()

# Display the Method Resolution Order (MRO) for both classes
print("\n" + "-" * 40 + "\n")
print("MFD_SPF Method Resolution Order:", [cls.__name__ for cls in MFD_SPF.__mro__])
print("MFD_SFP Method Resolution Order:", [cls.__name__ for cls in MFD_SFP.__mro__])
```
