## A static method in Python is a method inside a class that doesn’t depend on instance (self) or class (cls) data. It behaves like a regular function, but is logically grouped inside the class

```python
class MathUtils:
    
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(number):
        return number % 2 == 0


# Calling static methods
result = MathUtils.add(5, 3)
print(result)  # 8

print(MathUtils.is_even(4))  # True
```
