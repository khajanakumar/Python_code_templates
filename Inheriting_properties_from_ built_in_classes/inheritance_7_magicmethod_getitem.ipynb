```python
##  Example: Config Manager with Validation + Defaults + Logging

class ConfigManager:
    def __init__(self):
        self._config = {}
        self._defaults = {
            "theme": "light",
            "timeout": 30
        }

    def __setitem__(self, key, value):
        # Validation example
        if key == "timeout" and value < 0:
            raise ValueError("Timeout must be positive")

        print(f"[SET] {key} = {value}")
        self._config[key] = value

    def __getitem__(self, key):
        # Try user-defined config first
        if key in self._config:
            print(f"[GET] {key} (custom)")
            return self._config[key]

        # Fallback to defaults
        if key in self._defaults:
            print(f"[GET] {key} (default)")
            return self._defaults[key]

        # Key not found
        raise KeyError(f"{key} not found")


# Usage
config = ConfigManager()

# Setting values
config["theme"] = "dark"
config["timeout"] = 10

# Getting values
print(config["theme"])    # custom value
print(config["timeout"])  # custom value
print(config["language"]) # ❌ KeyError
```
