# This implementation includes:

1. **Abstract Classes**: `Scanner` and `Printer` abstract base classes with abstract methods
2. **Multiple Inheritance**: Each MFD inherits from both a scanner and printer class
3. **Method Overriding**: Each concrete class implements the abstract methods with specific behaviors
4. **Multiple Child Classes**: Three different MFD implementations with varying capabilities

```python
from abc import ABC, abstractmethod
from datetime import datetime

# Abstract Scanner class
class Scanner(ABC):
    @abstractmethod
    def scan_document(self):
        pass
    
    @abstractmethod
    def get_scanner_status(self):
        pass

# Abstract Printer class
class Printer(ABC):
    @abstractmethod
    def print_document(self):
        pass
    
    @abstractmethod
    def get_printer_status(self):
        pass

# Concrete Scanner implementations
class CheapScanner(Scanner):
    def __init__(self):
        self.resolution = "300 DPI"
        self.serial_number = "SCN-BASIC-" + str(id(self))[-4:]
    
    def scan_document(self):
        return f"Document scanned with basic quality ({self.resolution})"
    
    def get_scanner_status(self):
        return f"Scanner status: Resolution: {self.resolution}, S/N: {self.serial_number}"

class MediumScanner(Scanner):
    def __init__(self):
        self.resolution = "600 DPI"
        self.serial_number = "SCN-MID-" + str(id(self))[-4:]
    
    def scan_document(self):
        return f"Document scanned with good quality ({self.resolution})"
    
    def get_scanner_status(self):
        return f"Scanner status: Resolution: {self.resolution}, S/N: {self.serial_number}"

class PremiumScanner(Scanner):
    def __init__(self):
        self.resolution = "1200 DPI"
        self.serial_number = "SCN-PRO-" + str(id(self))[-4:]
    
    def scan_document(self):
        return f"Document scanned with premium quality ({self.resolution})"
    
    def get_scanner_status(self):
        return f"Scanner status: Resolution: {self.resolution}, S/N: {self.serial_number}"

# Concrete Printer implementations
class CheapPrinter(Printer):
    def __init__(self):
        self.resolution = "300 DPI"
        self.serial_number = "PRN-BASIC-" + str(id(self))[-4:]
    
    def print_document(self):
        return f"Document printed with basic quality ({self.resolution})"
    
    def get_printer_status(self):
        return f"Printer status: Resolution: {self.resolution}, S/N: {self.serial_number}"

class MediumPrinter(Printer):
    def __init__(self):
        self.resolution = "600 DPI"
        self.serial_number = "PRN-MID-" + str(id(self))[-4:]
        self.history = []
    
    def print_document(self):
        result = f"Document printed with good quality ({self.resolution})"
        self.history.append((datetime.now(), "print"))
        return result
    
    def get_printer_status(self):
        return f"Printer status: Resolution: {self.resolution}, S/N: {self.serial_number}"
    
    def print_operation_history(self):
        if not self.history:
            return "No operations in history"
        return "\n".join([f"{op[0]} - {op[1]}" for op in self.history])

class PremiumPrinter(Printer):
    def __init__(self):
        self.resolution = "1200 DPI"
        self.serial_number = "PRN-PRO-" + str(id(self))[-4:]
        self.history = []
    
    def print_document(self):
        result = f"Document printed with premium quality ({self.resolution})"
        self.history.append((datetime.now(), "print"))
        return result
    
    def get_printer_status(self):
        return f"Printer status: Resolution: {self.resolution}, S/N: {self.serial_number}"
    
    def print_operation_history(self):
        if not self.history:
            return "No operations in history"
        return "\n".join([f"{op[0]} - {op[1]}" for op in self.history])

# MFD implementations using multiple inheritance
class MFD1(CheapScanner, CheapPrinter):
    def __init__(self):
        CheapScanner.__init__(self)
        CheapPrinter.__init__(self)
        self.model = "EconoMFD-100"
    
    def __str__(self):
        return f"{self.model} - Budget Multifunction Device"

class MFD2(MediumScanner, MediumPrinter):
    def __init__(self):
        MediumScanner.__init__(self)
        MediumPrinter.__init__(self)
        self.model = "OfficeMFD-500"
    
    def __str__(self):
        return f"{self.model} - Standard Office Multifunction Device"
    
    def scan_document(self):
        result = super().scan_document()
        self.history.append((datetime.now(), "scan"))
        return result

class MFD3(PremiumScanner, PremiumPrinter):
    def __init__(self):
        PremiumScanner.__init__(self)
        PremiumPrinter.__init__(self)
        self.model = "ProMFD-9000"
        self.fax_number = "FAX-" + str(id(self))[-6:]
    
    def __str__(self):
        return f"{self.model} - Premium Multifunction Device with Fax"
    
    def scan_document(self):
        result = super().scan_document()
        self.history.append((datetime.now(), "scan"))
        return result
    
    def send_fax(self, recipient):
        result = f"Fax sent to {recipient} from {self.fax_number}"
        self.history.append((datetime.now(), f"fax to {recipient}"))
        return result

# Demonstration
if __name__ == "__main__":
    # Create instances of each MFD
    budget_mfd = MFD1()
    office_mfd = MFD2()
    premium_mfd = MFD3()
    
    # Test MFD1 (Budget)
    print("===== TESTING BUDGET MFD =====")
    print(budget_mfd)
    print(budget_mfd.scan_document())
    print(budget_mfd.print_document())
    print(budget_mfd.get_scanner_status())
    print(budget_mfd.get_printer_status())
    print()
    
    # Test MFD2 (Office)
    print("===== TESTING OFFICE MFD =====")
    print(office_mfd)
    print(office_mfd.scan_document())
    print(office_mfd.print_document())
    print(office_mfd.get_scanner_status())
    print(office_mfd.get_printer_status())
    print("Operation History:")
    print(office_mfd.print_operation_history())
    print()
    
    # Test MFD3 (Premium)
    print("===== TESTING PREMIUM MFD =====")
    print(premium_mfd)
    print(premium_mfd.scan_document())
    print(premium_mfd.print_document())
    print(premium_mfd.send_fax("123-456-7890"))
    print(premium_mfd.get_scanner_status())
    print(premium_mfd.get_printer_status())
    print("Operation History:")
    print(premium_mfd.print_operation_history())
```
