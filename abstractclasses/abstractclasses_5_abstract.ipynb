# Key Takeaways
- Use ABC to create abstract classes
- Use @abstractmethod to enforce method implementation
- Cannot instantiate abstract classes directly
- Helps enforce structure in object-oriented design

```python
from abc import ABC, abstractmethod

class Animal(ABC):  # Abstract class

    @abstractmethod
    def speak(self):
        pass


class Dog(Animal):
    def speak(self):
        return "Bark"


class Cat(Animal):
    def speak(self):
        return "Meow"


# Usage
dog = Dog()
cat = Cat()

print(dog.speak())  # Bark
print(cat.speak())  # Meow

# animal = Animal()  # ❌ TypeError: Can't instantiate abstract class
```

```python
## Example with Input Parameter
## Key Rule: The subclass method (Dog.speak) must be compatible with the abstract method signature
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self, value):
        pass


class Square(Shape):
    def area(self, side):
        return side * side


sq = Square()
print(sq.area(4))  # 16
```

```python
## Abstract + Concrete Methods Together
## No need to override concrete Methods (unless you want to)

from abc import ABC, abstractmethod

class Animal(ABC):

    def breathe(self):  # ✅ Concrete method (inherited as-is)
        return "Breathing..."

    @abstractmethod
    def speak(self):    # ❗ Must be implemented
        pass


class Dog(Animal):
    def speak(self):
        return "Bark"

dog = Dog()
print(dog.breathe())
print(dog.speak())

print (Dog.speak()) # Cannot call directly without instantiating
```
