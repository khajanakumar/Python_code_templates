```python
from abc import ABC, abstractmethod

class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass


class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass


class Duck(Flyable, Swimmable):  # Multiple inheritance
    def fly(self):
        return "Duck is flying"

    def swim(self):
        return "Duck is swimming"


# Usage
duck = Duck()

print(duck.fly())
print(duck.swim())
```

```python
from abc import ABC, abstractmethod

# Define the Flyable abstract class
class Flyable(ABC):
    def wings(self):
        return "Has wings"

    @abstractmethod
    def fly(self):
        pass

# Define the Swimmable abstract class that was missing
class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass

# Now Bird can inherit from both classes
class Bird(Flyable, Swimmable):
    def fly(self):
        return "Bird flying"

    def swim(self):
        return "Bird swimming"

# Create an instance and call methods
b = Bird()
print(b.fly())
print(b.wings())  # Inherited concrete method
```
