## What is `__class__`?

`__class__` is a special attribute (not exactly a magic method) that every Python object has. It provides a reference to the class that created the object instance. Think of it as a way for an object to "know" what type it is.

## Key Points:

1. **Instance to Class Reference**: `__class__` allows you to access the class of an instance
2. **Dynamic Access**: You can access class attributes and methods through an instance
3. **Type Checking**: Useful for determining object types at runtime
4. **Inheritance Aware**: Shows the actual class, not parent classes

Let me demonstrate with code:

```python
# Define a simple class hierarchy
class Animal:
    species_count = 0
    
    def __init__(self, name):
        self.name = name
        Animal.species_count += 1
    
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# Create instances
dog = Dog("Buddy")
cat = Cat("Whiskers")

# Using __class__ to access the class
print(f"Dog's class: {dog.__class__}")
print(f"Cat's class: {cat.__class__}")
print(f"Class name: {dog.__class__.__name__}")
print("*" * 20)


# Access class attributes through __class__
print(f"Species count: {dog.__class__.species_count}")
print("*" * 20)
# Create new instances using __class__
new_dog = dog.__class__("Rex")
print(f"New dog: {new_dog.name}, says: {new_dog.speak()}")
print(new_dog.__class__.__name__)
print("*" * 20)
# Compare with type() function
print(f"dog.__class__ == type(dog): {dog.__class__ == type(dog)}")

# Check inheritance
print(f"Is Dog a subclass of Animal? {issubclass(dog.__class__, Animal)}")

# Dynamic method calling
method_name = "speak"
if hasattr(dog.__class__, method_name):
    method = getattr(dog.__class__, method_name)
    print(f"Dynamic call: {method(dog)}")
```
