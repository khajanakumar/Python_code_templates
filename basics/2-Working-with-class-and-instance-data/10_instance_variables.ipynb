```python
class Student:
    # Class variable (shared by all objects)
    school_name = "ABC High School"

    def __init__(self, name, age):
        # Instance variables (unique to each object)
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, School: {Student.school_name}")


# Create objects
s1 = Student("Alice", 15)
s2 = Student("Bob", 16)

# Access instance variables
print(s1.name)  # Alice
print(s2.name)  # Bob

# Access class variable
print(Student.school_name)  # ABC High School

# Both objects share the same class variable
s1.display()
s2.display()

print("---- Modifying class variable ----")

# Modify class variable
Student.school_name = "XYZ School"

s1.display()
s2.display()

print("---- Modifying instance variable ----")

# Modify instance variable (only affects one object)
s1.name = "Charlie"

s1.display()
s2.display()
```

```python
class Student:
    school_name = "ABC School"  # class variable

    def __init__(self, name, age):
        self.name = name        # instance variable
        self.age = age          # instance variable


s1 = Student("Alice", 15)

# Check instance namespace
print(s1.__dict__)
print(Student.__dict__) 
```
