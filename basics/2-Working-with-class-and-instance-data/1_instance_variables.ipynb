Python allows for variables to be used at the instance level or the class level. Those used at the instance level are referred to as instance variables, whereas variables used at the class level are referred to as class variables.

Instance variables
This kind of variable exists when and only when it is explicitly created and added to an object. This can be done during the object's initialization, performed by the __init__ method, or later at any moment of the object's life. Furthermore, any existing property can be removed at any time.

Each object carries its own set of variables – they don't interfere with one another in any way. The word instance suggests that they are closely connected to the objects (which are class instances), not to the classes themselves. To get access to the instance variable, you should address the variable in the following way: objectdotvariable_name.

Let's look at how the instance variable is created and accessed in the code presented below.

class Demo:
    def __init__(self, value):
        self.instance_var = value

d1 = Demo(100)
d2 = Demo(200)

print("d1's instance variable is equal to:", d1.instance_var)
print("d2's instance variable is equal to:", d2.instance_var)


__init__ creates an instance_var variable for the instance. The keyword self is used to indicate that this variable is created coherently and individually for the instance to make it independent from other instances of the same class;

we instantiate the class twice, each time passing a different value to be stored inside the object;
the print instructions prove the fact that instance variable values are kept independently, because the printed values differ.

```python
class Demo:
    def __init__(self, value):
        self.instance_var = value

d1 = Demo(100)
d2 = Demo(200)

print("d1's instance variable is equal to:", d1.instance_var)
print("d2's instance variable is equal to:", d2.instance_var)

print (d1.__dict__)
print (d2.__dict__)
```

An instance variable in Python is initialized when an object (instance) of a class is created. Here are the key points about instance variable initialization:

## When Instance Variables Are Initialized:

1. **During object instantiation** - When you call the class constructor (e.g., `obj = MyClass()`)
2. **In the `__init__` method** - This is the most common place where instance variables are defined and given initial values
3. **Dynamically during runtime** - Instance variables can be created and initialized at any point after the object exists

## Common Initialization Patterns:

**Most Common: In `__init__` method**

```python
class Person:
    def __init__(self, name, age):
        self.name = name        # Initialized during object creation
        self.age = age          # Initialized during object creation
        self.email = None       # Initialized with default value

# Instance variables are initialized when object is created
person1 = Person("Alice", 30)
```

**Dynamic Initialization:**
- Instance variables can also be created and initialized later in other methods
- They can be added to an object at any time during its lifetime

## Key Points:
- Instance variables belong to a specific instance of the class
- Each object has its own copy of instance variables
- They are typically prefixed with `self.` when defined inside class methods
- Unlike class variables, instance variables are not shared between different objects of the same class

The initialization timing is crucial because it determines when the variable becomes available for use and what its initial state will be.
