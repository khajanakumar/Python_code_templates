```python
class Demo:
    class_var = 'shared variable'

d1 = Demo()
d2 = Demo()

print(Demo.class_var)
print(d1.class_var)
print(d2.class_var)
print ()
print('contents of d1:', d1.__dict__)
print()

Demo.class_var = 'Class shared variable'
print(Demo.class_var)
print(d1.class_var)
print(d2.class_var)
()
print(Demo.__dict__)
print(d1.__dict__)
print(d2.__dict__)
```

```python
d2.class_var = 'new shared variable'
print(d1.class_var)
print(d2.class_var)
print()
```
