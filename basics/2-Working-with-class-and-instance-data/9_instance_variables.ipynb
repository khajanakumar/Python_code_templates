# This solution uses:
- 1. A class variable `count` to track the number of apples processed
- 2. A class variable `total_weight` to track the total weight
- 3. Random weights between 0.2 and 0.5 for each apple
- 4. A packaging process that stops when either limit is reached


```python
import random

class Apple:
    # Class variables to track all apples
    count = 0
    total_weight = 0
    
    def __init__(self):
        # Generate random weight between 0.2 and 0.5
        self.weight = random.uniform(0.2, 0.5)
        
        # Update class variables
        Apple.count += 1
        Apple.total_weight += self.weight
    
    @classmethod
    def display_status(cls):
        """Display the current status of apple packaging"""
        print(f"Number of apples processed: {cls.count}")
        print(f"Total weight of apples: {cls.total_weight:.2f} units")

# Main packaging process
def package_apples(max_apples=1000, max_weight=300):
    print("Starting apple packaging process...")
    
    # Continue creating apples until a limitation is reached
    while Apple.count < max_apples and Apple.total_weight < max_weight:
        apple = Apple()
        
        # Optional: Uncomment to see details of each apple
        # print(f"Apple #{Apple.count}: weight = {apple.weight:.2f}")
        
        # Check if we've just exceeded the weight limit
        if Apple.total_weight > max_weight:
            break
    
    print("\nPackaging process completed!")
    print("Reason for stopping:", end=" ")
    
    if Apple.count >= max_apples:
        print("Maximum number of apples reached.")
    else:
        print("Maximum weight limit reached.")
    
    Apple.display_status()

# Run the packaging process
package_apples()
```
