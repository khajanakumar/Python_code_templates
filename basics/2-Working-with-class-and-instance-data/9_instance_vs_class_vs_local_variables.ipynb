## Types of Variables in a Class:

1. **Local variables** - Variables defined inside a method using regular assignment
2. **Instance variables** - Variables that belong to a specific instance, defined with `self.variable_name`
3. **Class variables** - Variables shared by all instances of the class

```python
class MyClass:
    class_var = "I'm a class variable"  # Class variable
    
    def __init__(self):
        self.instance_var = "I'm an instance variable"  # Instance variable
    
    def my_method(self):
        local_var = "I'm a local variable"  # Local variable (NOT instance variable)
        self.another_instance_var = "I'm also an instance variable. You can access from Instatiatated object"  # Instance variable
        
        print(f"Local variable: {local_var}")
        print(f"Instance variable: {self.instance_var}")
        print(f"Class variable: {MyClass.class_var}")

# Example usage
obj1 = MyClass()
obj2 = MyClass()

obj1.my_method()
# local_var is not accessible outside the method
# print(obj1.local_var)  # This would raise an AttributeError

# Instance variables are accessible
print(obj1.instance_var)  # Works fine
print(obj1.another_instance_var) # Can access the varable defined inside the function.
```

```python
print(obj1.__dict__)
print(obj2.__dict__) # you see only one variable as the my_method is not called
```
