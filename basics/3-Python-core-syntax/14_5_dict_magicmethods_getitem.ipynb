```python
## Basic Example: Custom Dictionary-like Class
class CustomDict:
    def __init__(self):
        self.data = {
            "name": "Alice",
            "age": 25
        }

    def __getitem__(self, key):
        print(f"Accessing key: {key}")
        return self.data[key]


cd = CustomDict()

print(cd["name"])  # Calls __getitem__
```

```python
## Example: Safe Access (No KeyError)
class SafeDict:
    def __init__(self):
        self.data = {"a": 1, "b": 2}

    def __getitem__(self, key):
        return self.data.get(key, "Key not found")


sd = SafeDict()

print(sd["a"])  # 1
print(sd["z"])  # "Key not found"
```

```python
# Example: Custom List Behavior

class DoubledList:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index] * 2


dl = DoubledList([1, 2, 3])

print(dl[0])  # 2
print(dl[1])  # 4
print (dl.items)
```
