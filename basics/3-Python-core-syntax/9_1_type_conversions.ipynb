Function	Magic method	Implementation meaning or purpose

- int()	    __int__(self)	conversion to integer type
- float()	__float__(self)	conversion to float type
- oct() 	__oct__(self)	conversion to string, containing an octal representation
- hex()	   __hex__(self)	conversion to string, containing a hexadecimal representation

```python
class Fraction:
    def __init__(self, numerator, denominator):
        if denominator == 0:
            raise ValueError("Denominator cannot be zero")
        self.numerator = numerator
        self.denominator = denominator
    
    def __float__(self):
        """Convert fraction to float by dividing numerator by denominator"""
        return self.numerator / self.denominator
    
    def __repr__(self):
        return f"Fraction({self.numerator}/{self.denominator})"

class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius
    
    def __float__(self):
        """Convert temperature to float (Celsius value)"""
        return float(self.celsius)
    
    def __repr__(self):
        return f"Temperature({self.celsius}°C)"

# Example usage
# Fraction example
frac1 = Fraction(3, 4)
print(f"Fraction: {frac1}")
print(f"As float: {float(frac1)}")  # Output: 0.75

frac2 = Fraction(22, 7)
print(f"Fraction: {frac2}")
print(f"As float: {float(frac2)}")  # Output: 3.142857142857143

# Temperature example
temp = Temperature(25)
print(f"Temperature: {temp}")
print(f"As float: {float(temp)}")  # Output: 25.0

# Using in mathematical operations
result = float(frac1) + float(frac2)
print(f"Sum: {result}")  # Output: 3.892857142857143
```
