Object introspection

Python offers a set of methods responsible for representing object details using ordinary strings.

Function	Magic method	Implementation meaning or purpose

- str()	    \__str__(self)	             responsible for handling str() function calls
- repr()	\__repr__(self)	             responsible for handling repr() function calls
- format()	\__format__(self, formatstr)	 called when new-style string formatting is applied to an object
- hash()	\__hash__(self)	             responsible for handling hash() function calls
- dir()	    \__dir__(self)	             responsible for handling dir() function calls
- bool()	\__nonzero__(self)	         responsible for handling bool() function calls

**Key Points:**
- The `__str__` method should return a string that provides a readable representation of the object
- It's automatically called when you use `print()` or `str()` on an object
- It should be designed for end-users (human-readable)
- Always return a string from this method

This makes your objects much more user-friendly when debugging or displaying information!

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"Person(name='{self.name}', age={self.age})"

# Create an instance
person = Person("Alice", 30)

# When we print the object, __str__ is automatically called
print(person)  # Output: Person(name='Alice', age=30)

# We can also call str() explicitly
print(str(person))  # Output: Person(name='Alice', age=30)

# Without __str__, the default representation would be something like:
# <__main__.Person object at 0x...>
```

**Key Differences:**

| Aspect | `__str__` | `__repr__` |
|--------|-----------|------------|
| **Purpose** | Human-readable, informal | Developer-friendly, unambiguous |
| **Audience** | End users | Developers/debugging |
| **Goal** | Readability | Reproducibility |
| **Fallback** | Uses `__repr__` if `__str__` not defined | No fallback |
| **Called by** | `print()`, `str()` | `repr()`, interactive shell, containers |
| **Ideal content** | Nice, readable format | Valid Python expression when possible |

**Best Practices:**
- `__repr__` should be unambiguous and ideally show how to recreate the object
- `__str__` should be readable and user-friendly
- Always implement `__repr__` first - it can serve as a fallback for `__str__`
- If you can only implement one, choose `__repr__`

The golden rule: `repr()` is for developers, `str()` is for users!

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
    
    def __str__(self):
        # Human-readable, informal representation
        return f"{self.title} by {self.author} ({self.year})"
    
    def __repr__(self):
        # Developer-friendly, unambiguous representation
        return f"Book(title='{self.title}', author='{self.author}', year={self.year})"

# Create instances
book1 = Book("1984", "George Orwell", 1949)
book2 = Book("To Kill a Mockingbird", "Harper Lee", 1960)

# Demonstrate __str__ (called by print())
print("Using print() - calls __str__:")
print(book1)  # Output: 1984 by George Orwell (1949)

# Demonstrate __repr__ (called by repr() or in interactive shell)
print("\nUsing repr() - calls __repr__:")
print(repr(book1))  # Output: Book(title='1984', author='George Orwell', year=1949)

# In a list, __repr__ is used for each element
print("\nIn a list - uses __repr__:")
books = [book1, book2]
print(books)  # Uses __repr__ for each book

# If __str__ is not defined, __repr__ is used as fallback
print("\nDirect object reference (in shell would show __repr__):")
book1  # In interactive shell, this would show the __repr__ output
```
