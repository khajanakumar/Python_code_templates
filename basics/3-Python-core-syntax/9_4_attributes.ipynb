```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Alice", 30)

# Basic usage
print(getattr(p, "name"))        # "Alice"
print(getattr(p, "age"))         # 30

# With a default (avoids AttributeError)
print(getattr(p, "email", "N/A"))  # "N/A" — no crash

# Without default, missing attr raises AttributeError
# getattr(p, "email")  # AttributeError: 'Person' object has no attribute 'email'

# Useful when attribute name is dynamic
field = "name"
print(getattr(p, field))  # "Alice"
```

```python
setattr(p, "age", 31)
print(p.age)  # 31

# Can create new attributes too
setattr(p, "email", "alice@example.com")
print(p.email)  # "alice@example.com"

# Dynamic field updates (e.g. from a dict)
updates = {"name": "Alicia", "age": 32}
for key, value in updates.items():
    setattr(p, key, value)

print(p.name, p.age)  # Alicia 32
```
