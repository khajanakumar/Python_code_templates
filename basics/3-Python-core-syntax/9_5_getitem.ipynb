# Methods allowing access to containers

Containers are any object that holds an arbitrary number of other objects; containers provide a way to access the contained objects and to iterate over them. Container examples: list, dictionary, tuple, and set.

Expression example	Magic method	Implementation meaning or purpose
- len(container)
    - \__len__(self)
    - returns the length (number of elements) of the container
- container[key]
    - \__getitem__(self, key)
    - responsible for accessing (fetching) an element identified by the key argument
- container[key] = value
    - \__setitem__(self, key, value)
    - responsible for setting a value to an element identified by the key argument
- del container[key]
    - \__delitem__(self, key)
    - responsible for deleting an element identified by the key argument
- for element in container
    - \__iter__(self)
    - returns an iterator for the container
- item in container
    - \__contains__(self, item)
    - responds to the question: does the container contain the selected item?

The list of special methods built-in in Python contains more entities. For more information, refer to https://docs.python.org/3/reference/datamodel.html#special-method-names.

Key points about `__getitem__`:

1. **Enables indexing**: Allows your object to be accessed with `obj[key]` syntax
2. **Supports slicing**: Python automatically handles slice objects when you use `obj[start:end]`
3. **Makes objects iterable**: If `__getitem__` is defined and raises `IndexError` when out of bounds, the object becomes iterable
4. **Flexible key types**: The `index` parameter can be integers, strings, tuples, or any hashable type depending on your implementation

This method is fundamental for creating container-like classes that behave similarly to built-in Python sequences like lists and tuples.

```python
# In this code, items is neither inherently a dictionary nor a list. 
# It's simply a parameter that can refer to any object that supports 
# indexing ([]) and len().
# If you pass a list sl = SimpleList([10, 20, 30]) the self.items refer to  [10,20,30]
# If you pass a dictionary sl = SimpleList({"a": 100, "b": 200}) 
# then self.items refer to {"a": 100, "b": 200}
# If you pass a string sl = SimpleList("Python") then self.item refers to "python"


class SimpleList:
    def __init__(self, items):
        self.items = items
    
    def __getitem__(self, index):
        """Enable indexing with [] operator"""
        return self.items[index]
    
    def __len__(self):
        """Enable len() function"""
        return len(self.items)

# Create an instance
my_list = SimpleList(['apple', 'banana', 'cherry', 'date'])

# Now we can use indexing
print(my_list[0])        # Output: apple
print(my_list[-1])       # Output: date
print(my_list[1:3])      # Output: ['banana', 'cherry'] (slicing works too!)
print(my_list[:]) 
# We can also iterate over it
for item in my_list:
    print(item)
```

Key points about using both methods together:

1. **`__setitem__`** enables assignment: `obj[key] = value`
2. **`__getitem__`** enables retrieval: `value = obj[key]`
3. **Together they make your class behave like a dictionary or list**
4. **Error handling**: You can customize what happens when keys don't exist
5. **Flexible implementation**: You can add logging, validation, or transformation logic

This combination allows you to create custom container classes that feel natural to use with Python's familiar indexing syntax.

```python
class SimpleDict:
    def __init__(self):
        self.data = {}
    
    def __getitem__(self, key):
        """Enable reading with [] operator"""
        if key not in self.data:
            raise KeyError(f"Key '{key}' not found")
        return self.data[key]
    
    def __setitem__(self, key, value):
        """Enable writing with [] operator"""
        self.data[key] = value
        print(f"Set {key} = {value}")
    
    def __len__(self):
        return len(self.data)
    
    def __repr__(self):
        return f"SimpleDict({self.data})"

# Create an instance
my_dict = SimpleDict()

# Use __setitem__ to add/modify values
my_dict['name'] = 'Alice'        # Output: Set name = Alice
my_dict['age'] = 30              # Output: Set age = 30
my_dict['city'] = 'New York'     # Output: Set city = New York

# Use __getitem__ to retrieve values
print(my_dict['name'])           # Output: Alice
print(my_dict['age'])            # Output: 30

# Modify existing value
my_dict['age'] = 31              # Output: Set age = 31
print(my_dict['age'])            # Output: 31

# Display the entire object
print(my_dict)                   # Output: SimpleDict({'name': 'Alice', 'age': 31, 'city': 'New York'})

# This will raise a KeyError
try:
    print(my_dict['country'])
except KeyError as e:
    print(e)                     # Output: Key 'country' not found
```

Key points about `__delitem__`:

1. **Enables `del` statement**: Allows `del obj[key]` syntax
2. **Error handling**: Should raise `KeyError` if the key doesn't exist (consistent with dict behavior)
3. **Cleanup logic**: You can add custom cleanup or logging when items are deleted
4. **Complete container interface**: Combined with `__getitem__` and `__setitem__`, your class now supports all basic container operations

The trio of `__getitem__`, `__setitem__`, and `__delitem__` makes your custom class behave like a built-in Python dictionary with full CRUD (Create, Read, Update, Delete) functionality.

```python
class SimpleDict:
    def __init__(self):
        self.data = {}
    
    def __getitem__(self, key):
        """Enable reading with [] operator"""
        if key not in self.data:
            raise KeyError(f"Key '{key}' not found")
        return self.data[key]
    
    def __setitem__(self, key, value):
        """Enable writing with [] operator"""
        self.data[key] = value
        print(f"Set {key} = {value}")
    
    def __delitem__(self, key):
        """Enable deletion with del statement"""
        if key not in self.data:
            raise KeyError(f"Key '{key}' not found")
        deleted_value = self.data[key]
        del self.data[key]
        print(f"Deleted {key} = {deleted_value}")
    
    def __len__(self):
        return len(self.data)
    
    def __repr__(self):
        return f"SimpleDict({self.data})"

# Create an instance
my_dict = SimpleDict()

# Use __setitem__ to add values
my_dict['name'] = 'Alice'        # Output: Set name = Alice
my_dict['age'] = 30              # Output: Set age = 30
my_dict['city'] = 'New York'     # Output: Set city = New York
my_dict['country'] = 'USA'       # Output: Set country = USA

print(my_dict)                   # Output: SimpleDict({'name': 'Alice', 'age': 30, 'city': 'New York', 'country': 'USA'})

# Use __getitem__ to retrieve values
print(f"Name: {my_dict['name']}")  # Output: Name: Alice

# Use __delitem__ to delete items
del my_dict['country']           # Output: Deleted country = USA
del my_dict['age']               # Output: Deleted age = 30

print(my_dict)                   # Output: SimpleDict({'name': 'Alice', 'city': 'New York'})

# Try to delete a non-existent key
try:
    del my_dict['phone']
except KeyError as e:
    print(f"Error: {e}")         # Output: Error: Key 'phone' not found

# Verify the remaining data
print(f"Remaining items: {len(my_dict)}")  # Output: Remaining items: 2
```
