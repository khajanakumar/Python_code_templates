```python

# Example of a composition
# Simple example: Car has an Engine

class Engine:
    def start(self):
        return "Engine started"


class Car:
    def __init__(self):
        # Composition: Car HAS an Engine
        self.engine = Engine()

    def start_car(self):
        return self.engine.start()


car = Car()
print(car.start_car())
```
