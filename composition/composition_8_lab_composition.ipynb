# Analysis of Vehicle Management System Code

This code demonstrates object-oriented programming in Python by implementing a vehicle management system with three classes: `Tires`, `Engine`, and `Vehicle`. The code showcases key OOP concepts like encapsulation, composition, and class relationships.

## Class Structure and Functionality

### 1. Tires Class
- **Purpose**: Manages tire properties and operations
- **Attributes**:
  - `size`: Tire diameter in inches
  - `pressure`: Current tire pressure (PSI), initialized to 0
- **Methods**:
  - `get_pressure()`: Returns the current tire pressure
  - `pump(amount)`: Increases tire pressure by the specified amount and prints the new pressure

### 2. Engine Class
- **Purpose**: Manages engine properties and operations
- **Attributes**:
  - `fuel_type`: Type of fuel the engine uses (e.g., "Electric", "Petrol")
  - `running`: Boolean indicating if the engine is running, initialized to False
- **Methods**:
  - `start()`: Starts the engine if not already running
  - `stop()`: Stops the engine if currently running
  - `get_state()`: Returns "running" or "stopped" based on engine state

### 3. Vehicle Class
- **Purpose**: Represents a complete vehicle that uses composition to incorporate engine and tires
- **Attributes**:
  - `VIN`: Vehicle Identification Number
  - `engine`: An Engine object
  - `tires`: A Tires object
- **Methods**:
  - `start_engine()`: Delegates to the engine's start method
  - `stop_engine()`: Delegates to the engine's stop method
  - `check_engine()`: Reports the engine's current state
  - `check_tire_pressure()`: Reports the current tire pressure
  - `pump_tires(amount)`: Delegates to the tires' pump method
  - `vehicle_info()`: Displays vehicle information including VIN, engine type, and tire size

## Demonstration Section

The code creates two different vehicle configurations:
1. **City Car**: Electric engine with 15-inch tires
2. **All-Terrain Car**: Petrol engine with 18-inch tires

It then demonstrates various operations on both vehicles:
- Displaying vehicle information
- Checking and adjusting tire pressure
- Starting and stopping engines
- Checking engine status

This code illustrates composition (a Vehicle "has-a" Engine and "has-a" set of Tires) rather than inheritance, which is often a more flexible design pattern for modeling real-world relationships between objects.

## Relevant Python Packages

For extending this type of object-oriented system, these Python packages might be useful:
- `dataclasses`: For creating classes with less boilerplate code
- `attrs`: For creating well-behaved classes with minimal code
- `pydantic`: For data validation and settings management
- `typing`: For adding type hints to improve code clarity and enable static type checking

```python
class Tires:
    def __init__(self, size):
        self.size = size
        self.pressure = 0
    
    def get_pressure(self):
        return self.pressure
    
    def pump(self, amount):
        self.pressure += amount
        print(f"Tires pumped to {self.pressure} PSI")


class Engine:
    def __init__(self, fuel_type):
        self.fuel_type = fuel_type
        self.running = False
    
    def start(self):
        if not self.running:
            self.running = True
            print(f"{self.fuel_type} engine started")
        else:
            print(f"{self.fuel_type} engine is already running")
    
    def stop(self):
        if self.running:
            self.running = False
            print(f"{self.fuel_type} engine stopped")
        else:
            print(f"{self.fuel_type} engine is already stopped")
    
    def get_state(self):
        return "running" if self.running else "stopped"


class Vehicle:
    def __init__(self, VIN, engine, tires):
        self.VIN = VIN
        self.engine = engine
        self.tires = tires
    
    def start_engine(self):
        self.engine.start()
    
    def stop_engine(self):
        self.engine.stop()
    
    def check_engine(self):
        state = self.engine.get_state()
        print(f"Engine is {state}")
    
    def check_tire_pressure(self):
        pressure = self.tires.get_pressure()
        print(f"Tire pressure is {pressure} PSI")
    
    def pump_tires(self, amount):
        self.tires.pump(amount)
    
    def vehicle_info(self):
        print(f"Vehicle VIN: {self.VIN}")
        print(f"Engine type: {self.engine.fuel_type}")
        print(f"Tire size: {self.tires.size} inches")


# Create tire sets
city_tires = Tires(size=15)
off_road_tires = Tires(size=18)

# Create engines
electric_engine = Engine(fuel_type="Electric")
petrol_engine = Engine(fuel_type="Petrol")

# Create vehicles
city_car = Vehicle(VIN="CITY123456", engine=electric_engine, tires=city_tires)
all_terrain_car = Vehicle(VIN="OFFRD789012", engine=petrol_engine, tires=off_road_tires)

# Demonstrate interactions with the city car
print("=== City Car Demo ===")
city_car.vehicle_info()
city_car.check_tire_pressure()
city_car.pump_tires(32)
city_car.check_tire_pressure()
city_car.start_engine()
city_car.check_engine()
city_car.stop_engine()

# Demonstrate interactions with the all-terrain car
print("\n=== All-Terrain Car Demo ===")
all_terrain_car.vehicle_info()
all_terrain_car.check_tire_pressure()
all_terrain_car.pump_tires(40)
all_terrain_car.start_engine()
all_terrain_car.check_engine()
all_terrain_car.stop_engine()
```
