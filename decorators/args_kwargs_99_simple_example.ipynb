# What are *args and **kwargs?
- *args → allows you to pass a variable number of positional arguments
- **kwargs → allows you to pass a variable number of keyword arguments
# Simple intuition
- *args → “I don’t know how many positional inputs you’ll give me”
- **kwargs → “I don’t know what named inputs you’ll give me”

```python
# args is a tuple and You can loop through it
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total


print(add_numbers(1, 2, 3))       # 6
print(add_numbers(10, 20, 30, 5)) # 65
```

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")


print_details(name="Alice", age=25, city="NY")
```

```python
# Using both together

def demo_function(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)


demo_function(1, 2, 3, name="Alice", age=30)
```

```python
# Mixing with regular parameters

def introduce(greeting, *args, **kwargs):
    print(greeting)

    for name in args:
        print(f"Hello {name}")

    for key, value in kwargs.items():
        print(f"{key}: {value}")


introduce("Welcome!", "Alice", "Bob", age=25, city="NY")
```

```python
# Unpacking arguments
def add(a, b, c):
    return a + b + c


numbers = [1, 2, 3]
print(add(*numbers))  # 6
```

```python
# Unpacking arguments
def greet(name, age):
    print(f"{name} is {age} years old")


data = {"name": "Alice", "age": 30}
greet(**data)
```
