```python
# Class decorator
def add_greeting(cls):
    cls.greet = lambda self: print("Hello! Welcome!")
    return cls


@add_greeting
class Student:
    def __init__(self, name):
        self.name = name


student = Student("Alice")

print(student.name)
student.greet()
```

```python
def add_school(cls):
    cls.school = "ABC High School"
    return cls


@add_school
class Student:
    pass


student = Student()

print(student.school)
```
