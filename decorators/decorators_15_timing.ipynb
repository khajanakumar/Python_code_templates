```python
from datetime import datetime
from functools import wraps

# Decorator that prints a timestamp
def log_timestamp(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        # Get current timestamp
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] Executing '{func.__name__}'")
        
        result = func(*args, **kwargs)
        
        return result
    return wrapper


# Example functions
@log_timestamp
def add(a, b):
    return a + b

@log_timestamp
def multiply(a, b):
    return a * b

@log_timestamp
def greet(name):
    print(f"Hello, {name}!")


# Test the functions
if __name__ == "__main__":
    print("Add result:", add(3, 5))
    print("Multiply result:", multiply(4, 6))
    greet("Alice")
```
