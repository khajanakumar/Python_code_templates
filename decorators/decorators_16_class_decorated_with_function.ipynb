```python
def count_objects(cls):
    cls.count = 0

    original_init = cls.__init__

    def new_init(self, *args, **kwargs):
        cls.count += 1
        original_init(self, *args, **kwargs)
        print(args)
        print(kwargs)

    cls.__init__ = new_init
    return cls


@count_objects
class Student:
    def __init__(self, name):
        self.name = name


s1 = Student("Alice")
s2 = Student("Bob")
s3 = Student("Charlie")

print(Student.count)
```
