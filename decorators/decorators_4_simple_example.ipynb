```python
# Suppose we want to print a message before and after every function call.
# Here's what's happening:
# func is the original function (greet)
# wrapper is a new function
# wrapper calls the original function and adds extra behavior
# The decorator returns wrapper

def my_decorator(func):

    def wrapper():
        print("Before the function runs")
        func()
        print("After the function runs")

    return wrapper
```

```python
# Step 3: Apply the decorator manually

def greet():
    print("Hello!")
greet = my_decorator(greet)
greet()
```

```python
# Step 4: Using the @ syntax
# Python provides a shortcut:

@my_decorator
def greet():
    print("Hello!")

greet()
```
