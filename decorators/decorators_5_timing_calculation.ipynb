```python
# Example: Timing a function
# Decorators are often used to measure execution time.

import time

def timer(func):

    def wrapper():
        start = time.time()

        func()

        end = time.time()
        print(f"Took {end-start:.3f} seconds")

    return wrapper
```

```python
@timer
def slow_function():
    time.sleep(2)
    print("Finished")

slow_function()
```
