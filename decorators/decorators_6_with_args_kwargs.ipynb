```python
# Decorators with arguements
# Most real functions accept parameters, so decorators usually use *args and **kwargs.
# *args collects positional arguments (5, 3).
# **kwargs collects keyword arguments (for example, name="Alice").
# Using both makes the decorator work with almost any function.

def my_decorator(func):

    def wrapper(*args, **kwargs):
        print("Calling", func.__name__)

        result = func(*args, **kwargs)

        print("Finished")
        return result

    return wrapper

@my_decorator
def add(a, b):
    return a + b

print(add(5, 3))
```

```python
print(add(5, 3,4))
```

```python
# A real-world example: Logging
# Suppose you want to log every function call.

def log_function(func):

    def wrapper(*args, **kwargs):
        print(f"Entering {func.__name__}")

        result = func(*args, **kwargs)

        print(f"Leaving {func.__name__}")

        return result

    return wrapper

@log_function
def multiply(x, y):
    return x * y

answer = multiply(4, 5)
print(answer)

```
