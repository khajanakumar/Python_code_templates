# What is a decorator?
- A decorator is a function that wraps another function to extend or modify its behavior without changing its code.

```python
# Functions are first-class objects
def greet():
    print("Hello!")

say_hello = greet
say_hello()  # Hello!
```

```python
# A function wrapping another function

def my_decorator(func):
    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")
    return wrapper
```

```python
def greet():
    print("Hello!")

greet = my_decorator(greet)

greet()
```

```python
# Using @ syntax (the real decorator)
@my_decorator
def greet():
    print("Hello!")

greet()
```

```python
# Logging decorator

def log_decorator(func):
    def wrapper():
        print(f"Calling function: {func.__name__}")
        func()
    return wrapper


@log_decorator
def say_hi():
    print("Hi!")

say_hi()
```

```python
# Decorator with arguments
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before execution")
        result = func(*args, **kwargs)
        print("After execution")
        return result
    return wrapper


@my_decorator
def add(a, b):
    return a + b


print(add(3, 4))
```

```python
# Real-world (authentication-style)

def require_login(func):
    def wrapper(user):
        if not user.get("is_logged_in"):
            print("Access denied")
            return
        return func(user)
    return wrapper


@require_login
def view_dashboard(user):
    print("Welcome to dashboard!")


user1 = {"is_logged_in": True}
user2 = {"is_logged_in": False}

view_dashboard(user1)  # Works
view_dashboard(user2)  # Access denied
```
