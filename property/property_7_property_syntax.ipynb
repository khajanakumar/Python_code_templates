## The same functionality could be implemented using the more modern decorator syntax

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius
 
    @property
    def radius(self):
        return self._radius
 
    @radius.setter
    def radius(self, value):
        if value > 0:
            self._radius = value
        else:
            raise ValueError("Radius must be a positive value.")

# Usage remains the same
circle = Circle(5)
circle.radius = 10
print(circle.radius)
```
