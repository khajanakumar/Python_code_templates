```python
class MyClass:
    def __init__(self):
        self.__internal = "This is internal data"
        self._protected = "This is protected data"
        self.public = "This is public data"

# Create an instance
obj = MyClass()

# Different ways to access variables
print(obj.public)                    # Works fine
print(obj._protected)                # Works but indicates "internal use"
# print(obj.__internal)              # This would raise AttributeError
print(obj._MyClass__internal)        # This works (accessing mangled name)

# See all attributes
print([attr for attr in dir(obj) if not attr.startswith('__')])
```
