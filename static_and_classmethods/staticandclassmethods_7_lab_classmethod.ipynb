```python
class LuxuryWatch:
    # Class variable to track number of watches created
    watches_created = 0

    def __init__(self):
        # Regular constructor does NOT allow engraving
        self.engraving = None
        LuxuryWatch.watches_created += 1

    @classmethod
    def with_engraving(cls, engraving_text):
        # Alternative constructor for engraved watches
        watch = cls()  # calls __init__ (increments counter)
        watch.engraving = engraving_text
        return watch

    @classmethod
    def get_number_of_watches_created(cls):
        return cls.watches_created


# --- Example usage ---

# Create regular watches
watch1 = LuxuryWatch()
watch2 = LuxuryWatch()

# Create engraved watch
watch3 = LuxuryWatch.with_engraving("To John, Stay Timeless")

print(LuxuryWatch.get_number_of_watches_created())  # Output: 3

print(watch1.engraving)  # Output: None
print(watch3.engraving)  # Output: To John, Stay Timeless
```
